2. Построим в плоскости графики полученных функций и нанесем экспериментальные точки

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([3, 5, 7, 9, 11, 13])
y = np.array([3.5, 4.4, 5.7, 6.1, 6.5, 7.3])

f1 = lambda t: 0.37 * t + 2.65
f2 = lambda t: 2.03 * t**0.50
f3 = lambda t: 3.09 * np.exp(0.07 * t)
f4 = lambda t: -0.02 * t**2 + 0.66 * t + 1.69

t = np.linspace(2.5, 13.5, 300)

plt.figure(figsize=(9, 6))
plt.scatter(x, y, color="black", zorder=5, label="Экспериментальные точки")
plt.plot(t, f1(t), label=r"$y_I = 0{,}37x + 2{,}65$")
plt.plot(t, f2(t), label=r"$y_{II} = 2{,}03\,x^{0{,}50}$")
plt.plot(t, f3(t), label=r"$y_{III} = 3{,}09\,e^{0{,}07x}$")
plt.plot(t, f4(t), label=r"$y_{IV} = -0{,}02x^2 + 0{,}66x + 1{,}69$")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.title("Графики аппроксимирующих функций и экспериментальные точки")
plt.grid(True)
plt.legend()
plt.show()
```
