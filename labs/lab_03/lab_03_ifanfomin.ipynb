Алгоритм распределения студентов по курсам:
1. По умолчанию на каждой дисциплине по выбору у 2 и 3 курсов может учиться 1 группа (до 30 студентов). Исключения описаны ниже. На blended-дисциплинах для третьекурсников количество мест не ограничено.
2. Проводится первая волна отбора. Для каждой дисциплины формируется список тех, кто указал её первым приоритетом (если студент должен выбрать два курса по выбору, то для него дисциплины, которые он указал первым и вторым приоритетом, рассматриваются как дисциплины первого приоритета). Если желающих больше, чем мест, то выбирается топ по перцентилю рейтинга.
3. На дисциплинах, где остались места после первой волны, формируются списки тех, кто выбрал их вторым приоритетом, и места заполняются лучшими по перцентили рейтинга студентами. После этого проводится такая же процедура для дисциплин третьего приоритета.
4. Если студент не попал на необходимое количество курсов по итогам трёх волн, с ним связывается учебный отдел и решает вопрос в индивидуальном порядке.

Распределите студентов в соответствии с первым приоритетом

Здесь для проверки приведена таблица, в которой есть 2 дополнительные колонки:
1) is_first_place - является ли студент лучшим по перцентили хотя бы на одном из курсов, куда он был зачислен 
(True / NaN)

2) is_last_place  - является ли студент худшим по перцентили хотя бы на одном из курсов, куда он был зачислен (True / NaN)

После распределения студентов в соответствии с первым приоритетом добавьте в свой датафрейм аналогичные признаки и запустите проверку:

Проведите все три волны отбора студентов на курсы по выбору Результат сохраните в файл res_fall.csv.

```python
check_df = pd.read_csv("./data/lab03_train_check.csv")

# Перцентиль студента из проверочной таблицы по его id
df["percentile"] = df.id.map(check_df.set_index("id").percentile)

columns = ["fall_1", "fall_2", "fall_3"]

# Получаю список всех уникальных осенних курсов, выбранных студентами
courses = np.unique(df[columns].stack().to_numpy())
fall_capacity = pd.Series(30.0, index=courses)

# Указываю курсы, для которых предусмотрена особая вместимость
special = {
        "Statistical Learning Theory": 60,
        "Высокопроизводительные вычисления": 60,
        "Анализ неструктурированных данных": np.inf,
}

# Заменяю стандартную вместимость на специальную для указанных курсов
for course, seats in special.items():
    fall_capacity[course] = seats

course_num = "fall_course_num"

# Одна строка будет соответствовать одному предпочтению студента
wave_1 = df.melt(
    id_vars=["id", "percentile", course_num],
    value_vars=columns,
    var_name="slot",
    value_name="course"
)

# Удаляю пропуски и повторный выбор одного курса студентом
wave_1 = wave_1.dropna(subset=["course"])
wave_1 = wave_1.sort_values(["id", "slot"])
wave_1 = wave_1.drop_duplicates(["id", "course"])

# Определяю номер волны для каждого предпочтения
priority = wave_1.slot.str[-1].astype(int)

wave_1["wave"] = np.where(
    wave_1[course_num] == 1,
    priority,
    (priority + 1) // 2
)

# Пока никто не зачислен
wave_1["admitted"] = False

# Сортирую по перцентилю: меньший перцентиль идёт первым
wave_1 = wave_1.sort_values(
    "percentile",
    ascending=True,
    kind="stable"
)

# Первая волна распределения
got = wave_1.groupby("id").admitted.transform("sum")
taken = wave_1.groupby("course").admitted.transform("sum")

seats_left = wave_1["course"].map(fall_capacity) - taken

candidates = wave_1[
    (wave_1["wave"] == 1) &
    (got < wave_1[course_num])
]

place = candidates.groupby("course").cumcount()

wave_1.loc[candidates.index, "admitted"] = (
    place < seats_left.loc[candidates.index]
)

# Выбираю зачисленных студентов
enrolled = wave_1[wave_1["admitted"]]

# Нахожу лучший и худший перцентиль среди зачисленных на каждый курс
best = enrolled.groupby("course").percentile.transform("min")
worst = enrolled.groupby("course").percentile.transform("max")

first_ids = enrolled.loc[
    enrolled.percentile == best, "id"
].unique()

last_ids = enrolled.loc[
    enrolled.percentile == worst, "id"
].unique()

# Формирую проверочные признаки
df["is_first_place"] = df["id"].isin(first_ids).map(
    {True: True, False: pd.NA}
)

df["is_last_place"] = df["id"].isin(last_ids).map(
    {True: True, False: pd.NA}
)

print("Выполнено")
score += 2
```
